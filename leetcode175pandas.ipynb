```python
'''175. Combine Two Tables'''
import pandas as pd

```

```python
def combine_two_tables(person: pd.DataFrame, address: pd.DataFrame) -> pd.DataFrame:
    combine=pd.merge(person,address,how='left').drop(columns=['personId','addressId'])
    return combine
```
